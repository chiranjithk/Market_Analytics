```python


# COMMAND ----------
from pyspark.sql import functions as F

stock_df = (
    spark.read.option("inferSchema", True).option("header", True)
    .csv(f"{MASTERS_VOLUME}/Equity_Stocks_List.csv")
    .withColumn("ingestion_timestamp", F.current_timestamp())
    .withColumnsRenamed({
        "SYMBOL": "Symbol",
        "NAME OF COMPANY": "Name",
        " SERIES": "Series",
        " DATE OF LISTING": "Listing_date",
        " PAID UP VALUE": "Paid_up_value",
        " FACE VALUE": "Face_value",
        " MARKET LOT": "Market_lot",
        " ISIN NUMBER": "isin_number",
    })
)

stock_df.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE}.stock_master_raw")
display(stock_df)
```
